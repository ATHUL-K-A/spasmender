```python
import pandas as pd

CSV_FILE     = "D:\\Project\\Dataset EMG Fatigue\\candy_read\\data\\Athul\\TL50.csv"
NONFAT_END   = 20.0    # first 20 seconds = non-fatigued
FATIGUED_START = 40.0  # after n seconds = fatigued (discard 30-50s as transition)

df = pd.read_csv(CSV_FILE)

nonfatigued = df[df["Time_s"] <= NONFAT_END].copy()
fatigued    = df[df["Time_s"] >= FATIGUED_START].copy()

nonfatigued["label"] = 0
fatigued["label"]    = 1

nonfatigued.to_csv("D:\\Project\\Dataset EMG Fatigue\\candy_read\\data\\Athul\\session1_nonfatiguedTL_50.csv", index=False)
fatigued.to_csv("D:\\Project\\Dataset EMG Fatigue\\candy_read\\data\\Athul\\session1_fatiguedTL_50.csv",    index=False)

print(f"Non-fatigued samples: {len(nonfatigued)}")
print(f"Fatigued samples:     {len(fatigued)}")
```
